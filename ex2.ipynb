## Q1

```python
import numpy as np

arr = np.arange(1, 13)  

arr_2d = arr.reshape(3, 4)

print("数组的维度(ndim):", arr_2d.ndim)
print("数组的形状(shape):", arr_2d.shape)
print("数组的元素总个数(size):", arr_2d.size)
```

## Q2

```python
import numpy as np

arr = np.array([1.5, 2.6, 3.7, 4.8])

print("转换前的数组：", arr)
print("转换前的数据类型：", arr.dtype.name)

arr_int = arr.astype(np.int32)

print("转换后的数组：", arr_int)
print("转换后的数据类型：", arr_int.dtype.name)
```

## Q4

```python
import numpy as np

arr = np.array([10, 15, 20, 25, 30, 35])

fancy = arr[[0, 2, 4]]
print("花式索引结果：", fancy)

bool_idx = arr > 20
bool_res = arr[bool_idx]
print("布尔索引（大于20）：", bool_res)
```

## Q5

```python
import numpy as np

arr_a = np.zeros((3, 4))
arr_b = np.array([1, 2, 3, 4])

result = arr_a + arr_b
print("广播相加结果：")
print(result)
```

## Q6

```python
import numpy as np

x = np.array([1, 4, 9, 16])
y = np.array([2, 1, 4, 5])

sqrt_x = np.sqrt(x)
print("x的平方根：", sqrt_x)

max_xy = np.maximum(sqrt_x, y)
print("对应位置最大值：", max_xy)
```

## Q8

```python
import numpy as np

arr_x = np.array([1, 5, 7])
arr_y = np.array([2, 6, 8])

cond = np.array([True, False, True])

result = np.where(cond, arr_x, arr_y)
print("条件逻辑运算结果：", result)
```

## Q9

```python
import numpy as np

arr = np.array([[6, 2, 7], [3, 6, 2], [2, 4, 3]])

mean_all = arr.mean()
print("所有元素平均值：", mean_all)

arr_sorted = np.sort(arr, axis=0)
print("沿axis=0排序后的数组：")
print(arr_sorted)
```

## Q10

```python
import numpy as np

np.random.seed(10)

A = np.random.rand(2, 2)
B = np.random.rand(2, 2)

dot_result = np.dot(A, B)

print("矩阵A：")
print(A)
print("\n矩阵B：")
print(B)
print("\nA·B 矩阵点积：")
print(dot_result)
```
