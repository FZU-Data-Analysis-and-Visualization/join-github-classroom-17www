## Q2

```python
# ====================== 二、列与行操作 ======================
import pandas as pd
dict_data = {
    "name": ["Tom", "Jerry", "Mike"],
    "score": [85, 92, 78],
    "gender": ["男", "女", "男"]
}
df = pd.DataFrame(dict_data)

#1.选取score、gender两列
col_df = df[["score", "gender"]]
print("选取两列：\n", col_df)

#2.选取索引0、2行
row_df = df.loc[[0, 2]]
print("\n选取索引0、2行：\n", row_df)

#3.筛选score>80
filter_df = df[df["score"] > 80]
print("\nscore>80：\n", filter_df)

#4.新增grade列
df["grade"] = df["score"].apply(lambda x: "优秀" if x >=90 else "良好")
print("\n新增grade：\n", df)

#5.删除gender列
drop_df = df.drop("gender", axis=1)
print("\n删除gender：\n", drop_df)
```

```python
# ====================== 三、缺失值 ======================
import pandas as pd
#1.创建缺失值df，检测+统计缺失
miss_data = {"x": [1, None, 3], "y": [None, 5, 6]}
df_miss = pd.DataFrame(miss_data)
print("缺失值df:\n", df_miss)
print("缺失布尔：\n", df_miss.isna())
print("每列缺失数量：\n", df_miss.isna().sum())

#2.0填充；删除缺失行
fill0 = df_miss.fillna(0)
drop_na = df_miss.dropna()
print("\n0填充：\n", fill0)
print("删除缺行：\n", drop_na)
```

## Q5

```python
# ====================== 五、排序与去重 ======================
import pandas as pd
dict_data = {
    "name": ["Tom", "Jerry", "Mike"],
    "score": [85, 92, 78],
    "gender": ["男", "女", "男"]
}
df = pd.DataFrame(dict_data)

#1.score升序
sort_df = df.sort_values("score", ascending=True)
print("升序排序：\n", sort_df)

#2.添加重复行后去重
add_row = pd.DataFrame([["Tom",85,"男"]], columns=df.columns)
df_add = pd.concat([df, add_row], ignore_index=True)
df_dropdup = df_add.drop_duplicates()
print("\n添加重复后：\n", df_add)
print("去重后：\n", df_dropdup)
```

```python
# ====================== 六、合并与拼接 ======================
import pandas as pd
#1.横向合并(按id)
df1 = pd.DataFrame({"id":[1,2], "name":["X","Y"]})
df2 = pd.DataFrame({"id":[1,2], "age":[20,30]})
merge_df = pd.merge(df1, df2, on="id")
print("按id合并：\n", merge_df)

#2.纵向拼接+重置索引
df_a = pd.DataFrame({"val":[1,2]})
df_b = pd.DataFrame({"val":[3,4]})
concat_df = pd.concat([df_a, df_b], ignore_index=True)
print("\n纵向拼接：\n", concat_df)
```

## Q7

```python
# ====================== 七、时间处理 ======================
import pandas as pd
#1.转为datetime
date_list = ["2025-01-01","2025-01-02","2025-01-03"]
dt_ser = pd.to_datetime(date_list)
print("时间列：\n", dt_ser)

#2.提取月份、星期
df_dt = pd.DataFrame({"date":dt_ser})
df_dt["month"] = df_dt["date"].dt.month
df_dt["weekday"] = df_dt["date"].dt.day_name()
print("\n提取月份星期：\n", df_dt)
```

## Q8

```python
# ====================== 八、高级操作 ======================
import pandas as pd
dict_data = {
    "name": ["Tom", "Jerry", "Mike"],
    "score": [85, 92, 78],
    "gender": ["男", "女", "男"]
}
df = pd.DataFrame(dict_data)

#1.score标准化 (x-mean)/std
df["score_norm"] = (df["score"] - df["score"].mean()) / df["score"].std()
print("标准化：\n", df)

#2.透视表：行gender，值score，聚合中位数
pivot_df = pd.pivot_table(df, index="gender", values="score", aggfunc="median")
print("\n透视表中位数：\n", pivot_df)
```
